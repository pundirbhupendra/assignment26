## Q3. Connect MongoDB to Python and create database and collection.

```
from pymongo import MongoClient

# Connect MongoDB
client = MongoClient("mongodb://localhost:27017/")

print("MongoDB connected successfully")

# Create database
db = client["pwskills"]

# Create collection
collection = db["students"]

print("Database and Collection created successfully")
```

## Q4. Insert one and many records and print them.

```
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["pwskills"]
collection = db["students"]

# Insert one record
collection.insert_one({
    "name": "John",
    "age": 25
})

# Insert many records
collection.insert_many([
    {"name": "Alice", "age": 22},
    {"name": "Bob", "age": 24}
])

print("One record using find_one():")
print(collection.find_one())

print("\nAll records using find():")
for record in collection.find():
    print(record)
```

## Q5. Explain find() method.

```
print("find() method is used to query and retrieve documents from a MongoDB collection.")

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["pwskills"]
collection = db["students"]

print("\nStudents with age greater than 22:")

for record in collection.find({"age": {"$gt": 22}}):
    print(record)
```

## Q6. Explain sort() method.

```
print("sort() method is used to sort documents in ascending or descending order.")

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["pwskills"]
collection = db["students"]

print("\nRecords sorted by age in ascending order:")

for record in collection.find().sort("age", 1):
    print(record)
```

## Q7. Explain delete_one(), delete_many(), and drop().

```
print("delete_one() -> Deletes a single document matching the condition.")
print("delete_many() -> Deletes multiple documents matching the condition.")
print("drop() -> Deletes the entire collection.")

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["pwskills"]
collection = db["students"]

# delete_one example
collection.delete_one({"name": "John"})

# delete_many example
collection.delete_many({"age": {"$lt": 24}})

# drop example
# collection.drop()

print("Deletion operations executed successfully")
```
