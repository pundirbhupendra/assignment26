```python
## Q1. What is an activation function in the context of artificial neural networks?

An **activation function** is a mathematical function used in a neural network to decide whether a neuron should be activated or not. It takes the input received by a neuron and transforms it into an output that is passed to the next layer.

Activation functions help neural networks learn **complex and non-linear patterns** in data.

Example:
Without activation functions, a neural network would behave like a simple linear model and could not solve complex problems like image recognition or language processing.

---

## Q2. What are some common types of activation functions used in neural networks?

Some commonly used activation functions are:

1. **Sigmoid Function**
2. **Tanh (Hyperbolic Tangent)**
3. **ReLU (Rectified Linear Unit)**
4. **Leaky ReLU**
5. **Softmax**
6. **ELU (Exponential Linear Unit)**

These functions are chosen depending on the task and neural network architecture.

---

## Q3. How do activation functions affect the training process and performance of a neural network?

Activation functions affect:

* **Learning capability** of the network
* **Training speed**
* **Gradient flow during backpropagation**
* **Accuracy and performance**

Key effects:

* They introduce **non-linearity**, allowing the network to solve complex tasks.
* Some activation functions may cause problems like:

  * **Vanishing gradient problem**
  * **Dead neurons**
* Proper activation functions improve convergence and model performance.

---

## Q4. How does the sigmoid activation function work? What are its advantages and disadvantages?

The sigmoid function converts input values into outputs between **0 and 1**.

\sigma(x)=\frac{1}{1+e^{-x}}

### Advantages

* Output values are between 0 and 1
* Useful for probability-based outputs
* Commonly used in binary classification

### Disadvantages

* Causes **vanishing gradient problem**
* Training becomes slow for deep networks
* Output is not zero-centered

---

## Q5. What is the rectified linear unit (ReLU) activation function? How does it differ from the sigmoid function?

ReLU is one of the most widely used activation functions in deep learning.

f(x)=\max(0,x)

### Working

* If input is positive → output is same value
* If input is negative → output becomes 0

### Difference between ReLU and Sigmoid

| ReLU                          | Sigmoid                         |
| ----------------------------- | ------------------------------- |
| Output range: 0 to ∞          | Output range: 0 to 1            |
| Faster training               | Slower training                 |
| Less vanishing gradient issue | Suffers from vanishing gradient |
| Simple computation            | More computationally expensive  |

---

## Q6. What are the benefits of using the ReLU activation function over the sigmoid function?

Benefits of ReLU:

1. **Faster computation**
2. **Speeds up training**
3. **Reduces vanishing gradient problem**
4. **Works well in deep neural networks**
5. **Simple mathematical operation**

Because of these advantages, ReLU is commonly used in modern deep learning models.

---

## Q7. Explain the concept of "leaky ReLU" and how it addresses the vanishing gradient problem.

Leaky ReLU is an improved version of ReLU.

f(x)=\begin{cases}x,&x>0\0.01x,&x\le0\end{cases}

### Problem with ReLU

In ReLU, negative inputs produce 0 output, causing some neurons to stop learning permanently. This is called the **dying ReLU problem**.

### How Leaky ReLU Helps

Leaky ReLU allows a small negative output for negative inputs instead of 0.

Benefits:

* Keeps gradients flowing
* Reduces dying neurons
* Improves learning in deep networks

---

## Q8. What is the purpose of the softmax activation function? When is it commonly used?

The softmax function converts outputs into **probability values** whose sum equals 1.

\text{Softmax}(x_i)=\frac{e^{x_i}}{\sum_{j}e^{x_j}}

### Purpose

* Used for **multi-class classification**
* Determines probability of each class

### Common Uses

* Image classification
* Digit recognition
* NLP classification tasks

Example:
If a model predicts cat = 0.7, dog = 0.2, bird = 0.1, softmax ensures total probability equals 1.

---

## Q9. What is the hyperbolic tangent (tanh) activation function? How does it compare to the sigmoid function?

The tanh activation function outputs values between **-1 and 1**.

\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}

### Advantages of tanh

* Zero-centered output
* Better convergence than sigmoid in many cases

### Comparison with Sigmoid

| Tanh                                   | Sigmoid                               |
| -------------------------------------- | ------------------------------------- |
| Output range: -1 to 1                  | Output range: 0 to 1                  |
| Zero-centered                          | Not zero-centered                     |
| Often performs better in hidden layers | Mostly used in output layers          |
| Still suffers from vanishing gradients | Also suffers from vanishing gradients |
```
