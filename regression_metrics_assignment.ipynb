# Regression Metrics and Regularization Assignment

## Q1. Explain the concept of R-squared in linear regression

R-squared (R²) measures how well the regression model explains the variation in the dependent variable.

### Formula:
R² = 1 - (SSR / SST)

Where:
- SSR = Sum of Squared Residuals
- SST = Total Sum of Squares

### Interpretation:
- R² = 1 → Perfect fit
- R² = 0 → No explanatory power

Example:
If R² = 0.90, then 90% of the variation in the target variable is explained by the model.

## Q2. Define adjusted R-squared

Adjusted R-squared is a modified form of R² that adjusts for the number of independent variables.

### Difference:
- R² always increases when variables are added.
- Adjusted R² increases only if the added variable improves the model.

It helps avoid overestimating model performance.

## Q3. When is adjusted R-squared more appropriate?

Adjusted R² is more appropriate in multiple linear regression when multiple predictors are used.

It helps evaluate whether adding new features actually improves the model.

## Q4. Explain RMSE, MSE, and MAE

### MSE (Mean Squared Error)
Average squared difference between actual and predicted values.

### RMSE (Root Mean Squared Error)
Square root of MSE.

### MAE (Mean Absolute Error)
Average absolute difference between actual and predicted values.

### Interpretation:
Lower values indicate better model performance.

## Q5. Advantages and disadvantages of RMSE, MSE, and MAE

### MSE
Advantages:
- Penalizes large errors heavily

Disadvantages:
- Sensitive to outliers

### RMSE
Advantages:
- Easy to interpret
- Same unit as target variable

Disadvantages:
- Sensitive to outliers

### MAE
Advantages:
- Less sensitive to outliers
- Simple interpretation

Disadvantages:
- Does not penalize large errors strongly

## Q6. Explain Lasso regularization

Lasso regularization uses L1 penalty.

### Formula:
Loss + λΣ|β|

### Difference from Ridge:
- Ridge uses L2 penalty
- Lasso can reduce coefficients to zero

### Usage:
Lasso is useful for feature selection.

## Q7. How regularized models prevent overfitting

Regularization adds penalties to large coefficients, preventing the model from fitting noise in the training data.

This improves model generalization on unseen data.

## Q8. Limitations of regularized linear models

Limitations:
1. Can underfit data
2. Assumes linear relationships
3. Sensitive to feature scaling
4. May remove important variables

Regularized models may not work well for highly nonlinear datasets.

## Q9. Comparing Model A and Model B

It is difficult to compare RMSE and MAE directly because they measure errors differently.

### RMSE:
- Penalizes large errors more

### MAE:
- Treats all errors equally

Without evaluating both metrics for both models, a final conclusion cannot be made.

## Q10. Comparing Ridge and Lasso models

### Ridge:
- Keeps all features
- Reduces coefficient values

### Lasso:
- Performs feature selection
- Removes irrelevant features

### Choice:
- Use Ridge when all variables are important.
- Use Lasso when feature selection is needed.

### Trade-offs:
Lasso simplifies models, while Ridge improves stability.
