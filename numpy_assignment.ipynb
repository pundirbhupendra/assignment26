# NumPy Assignment

## Given Code

```
import numpy as np

list_ = ['1', '2', '3', '4', '5']

array_list = np.array(object=list_)

print("List:", list_)
print("Array:", array_list)
```

## Q1. Difference in data types of list_ and array_list.

```
print("Type of list_:", type(list_))
print("Type of array_list:", type(array_list))
```

## Q2. Print data type of each element.

```
print("Data types in list_:")
for i in list_:
    print(type(i))

print("\nData types in array_list:")
for i in array_list:
    print(type(i))
```

## Q3. Change dtype to int and check element types.

```
array_list = np.array(object=list_, dtype=int)

print("Data types in list_:")
for i in list_:
    print(type(i))

print("\nData types in array_list:")
for i in array_list:
    print(type(i))
```

## Q4. Find shape and size of num_array.

```
num_list = [[1,2,3],[4,5,6]]

num_array = np.array(object=num_list)

print("Shape:", num_array.shape)
print("Size:", num_array.size)
```

## Q5. Create 3x3 matrix containing zeros only.

```
zero_matrix = np.zeros((3,3))

print(zero_matrix)
```

## Q6. Create identity matrix of shape (5,5).

```
identity_matrix = np.eye(5)

print(identity_matrix)
```
