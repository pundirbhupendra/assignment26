## Given DataFrame

```
import pandas as pd

course_name = ['Data Science', 'Machine Learning', 'Big Data', 'Data Engineer']
duration = [2, 3, 6, 4]

df = pd.DataFrame({
    'course_name': course_name,
    'duration': duration
})

print(df)
```

## Q1. Print the data present in the second row of the dataframe.

```
print(df.iloc[1])
```

## Q2. Difference between loc and iloc.

```
print("loc -> label based indexing")
print("iloc -> integer position based indexing")
```

## Q3. Reindex dataframe and find outputs.

```
new_df = df.reindex([3,0,1,2])

print("new_df:")
print(new_df)

print("\nnew_df.loc[2]:")
print(new_df.loc[2])

print("\nnew_df.iloc[2]:")
print(new_df.iloc[2])
```

## Q4. Statistical measurements.

```
import numpy as np

df1 = pd.DataFrame({
    'column_1': [1,2,3,4],
    'column_2': [10,20,30,40]
})

print("Mean of each column:")
print(df1.mean())

print("\nStandard deviation of column_2:")
print(df1['column_2'].std())
```

## Q5. Replace second row data by string and find mean.

```
df1.loc[1, 'column_2'] = 'PW'

print(df1)

print("\nTrying to calculate mean:")
try:
    print(df1['column_2'].mean())
except Exception as e:
    print("Error:", e)

print("\nExplanation: Mean cannot be calculated because column_2 now contains string data.")
```

## Q8. Difference between two dates.

```
date1 = input("Enter first date (YYYY-MM-DD): ")
date2 = input("Enter second date (YYYY-MM-DD): ")

d1 = pd.to_datetime(date1)
d2 = pd.to_datetime(date2)

difference = d2 - d1

print("Difference:", difference)
print("Days:", difference.days)
print("Hours:", difference.seconds // 3600)
print("Minutes:", (difference.seconds % 3600) // 60)
```

## Q9. Convert column to categorical type.

```
file_path = input("Enter CSV file path: ")
column_name = input("Enter column name: ")
order = input("Enter categories separated by commas: ").split(',')

df_cat = pd.read_csv(file_path)

df_cat[column_name] = pd.Categorical(df_cat[column_name], categories=order, ordered=True)

sorted_df = df_cat.sort_values(column_name)

print(sorted_df)
```

## Q10. Visualize sales data using stacked bar chart.

```
import matplotlib.pyplot as plt

file_path = input("Enter CSV file path: ")

sales_df = pd.read_csv(file_path)

sales_df.plot(kind='bar', stacked=True)

plt.title("Sales Data")
plt.xlabel("Index")
plt.ylabel("Sales")
plt.show()
```

## Q11. Calculate mean, median, and mode from student CSV.

```
file_path = input("Enter CSV file path: ")

student_df = pd.read_csv(file_path)

mean_score = student_df['Test Score'].mean()
median_score = student_df['Test Score'].median()
mode_score = student_df['Test Score'].mode().tolist()

result = pd.DataFrame({
    'Statistic': ['Mean', 'Median', 'Mode'],
    'Value': [mean_score, median_score, mode_score]
})

print(result)
```
