## Q1. What is a database? Differentiate between SQL and NoSQL databases.

```
print("Database: A database is an organized collection of data that can be stored, managed, and retrieved efficiently.")

print("\nDifference between SQL and NoSQL Databases:")

print("1. SQL databases are relational, while NoSQL databases are non-relational.")
print("2. SQL uses tables, NoSQL uses documents, key-value pairs, graphs, etc.")
print("3. SQL databases follow ACID properties.")
print("4. NoSQL databases are more scalable for large unstructured data.")
print("5. Examples of SQL: MySQL, PostgreSQL")
print("6. Examples of NoSQL: MongoDB, Cassandra")
```

## Q2. What is DDL? Explain CREATE, DROP, ALTER, and TRUNCATE.

```
print("DDL stands for Data Definition Language.")

print("\nCREATE Example:")
print("CREATE TABLE students(id INT, name VARCHAR(50));")

print("\nDROP Example:")
print("DROP TABLE students;")

print("\nALTER Example:")
print("ALTER TABLE students ADD age INT;")

print("\nTRUNCATE Example:")
print("TRUNCATE TABLE students;")
```

## Q3. What is DML? Explain INSERT, UPDATE, and DELETE.

```
print("DML stands for Data Manipulation Language.")

print("\nINSERT Example:")
print("INSERT INTO students VALUES(1, 'John');")

print("\nUPDATE Example:")
print("UPDATE students SET name='Sam' WHERE id=1;")

print("\nDELETE Example:")
print("DELETE FROM students WHERE id=1;")
```

## Q4. What is DQL? Explain SELECT.

```
print("DQL stands for Data Query Language.")

print("\nSELECT Example:")
print("SELECT * FROM students;")
```

## Q5. Explain Primary Key and Foreign Key.

```
print("Primary Key: A column that uniquely identifies each row in a table.")
print("Example: student_id in students table")

print("\nForeign Key: A column that creates a relationship between two tables.")
print("Example: department_id in students table referring to departments table")
```

## Q6. Python code to connect MySQL to Python.

```
import mysql.connector

# Connection example
mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    password="password",
    database="testdb"
)

print("Database connected successfully")

# cursor() method
cursor = mydb.cursor()

print("\ncursor() method creates a cursor object to execute SQL queries.")

# execute() method
cursor.execute("SHOW TABLES")

print("\nexecute() method is used to execute SQL queries.")
```
