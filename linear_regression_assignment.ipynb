# Linear Regression Assignment

## Q1. Difference between simple linear regression and multiple linear regression

### Simple Linear Regression
Simple linear regression uses one independent variable to predict a dependent variable.

Example:
Predicting house price based only on house size.

Equation:
y = b0 + b1x

### Multiple Linear Regression
Multiple linear regression uses multiple independent variables to predict a dependent variable.

Example:
Predicting house price based on house size, number of bedrooms, and location.

Equation:
y = b0 + b1x1 + b2x2 + b3x3 + ...

## Q2. Assumptions of linear regression

The assumptions of linear regression are:

1. Linearity:
   Relationship between independent and dependent variables should be linear.

2. Independence:
   Observations should be independent.

3. Homoscedasticity:
   Constant variance of errors.

4. Normality:
   Residuals should be normally distributed.

5. No multicollinearity:
   Independent variables should not be highly correlated.

### Checking assumptions:
- Scatter plots for linearity
- Residual plots for homoscedasticity
- Histogram or Q-Q plot for normality
- VIF (Variance Inflation Factor) for multicollinearity

## Q3. Interpretation of slope and intercept

### Slope
The slope represents the change in the dependent variable for a one-unit increase in the independent variable.

### Intercept
The intercept is the value of the dependent variable when the independent variable is zero.

### Example
Suppose:
Salary = 3000 + 500 × Experience

- Intercept (3000):
  Starting salary with zero years of experience.

- Slope (500):
  Salary increases by 500 for every additional year of experience.

## Q4. Concept of gradient descent

Gradient descent is an optimization algorithm used to minimize the loss function by updating model parameters iteratively.

### Steps:
1. Initialize parameters randomly
2. Compute loss
3. Calculate gradients
4. Update parameters using learning rate
5. Repeat until convergence

### Use in Machine Learning
Gradient descent is used to train machine learning models such as:
- Linear regression
- Logistic regression
- Neural networks

## Q5. Multiple linear regression model

Multiple linear regression predicts a dependent variable using multiple independent variables.

### Equation:
y = b0 + b1x1 + b2x2 + ... + bnxn

### Difference from simple linear regression:
- Simple linear regression uses one independent variable.
- Multiple linear regression uses multiple independent variables.

## Q6. Multicollinearity in multiple linear regression

Multicollinearity occurs when independent variables are highly correlated with each other.

### Problems caused:
- Unstable coefficient estimates
- Reduced model interpretability

### Detection methods:
- Correlation matrix
- Variance Inflation Factor (VIF)

### Solutions:
- Remove correlated variables
- Apply dimensionality reduction techniques
- Use regularization methods

## Q7. Polynomial regression model

Polynomial regression models nonlinear relationships between variables.

### Equation:
y = b0 + b1x + b2x² + b3x³ + ...

### Difference from linear regression:
- Linear regression fits a straight line.
- Polynomial regression fits a curved relationship.

## Q8. Advantages and disadvantages of polynomial regression

### Advantages:
1. Captures nonlinear relationships
2. More flexible than linear regression
3. Better fit for complex datasets

### Disadvantages:
1. Can overfit the data
2. Computationally expensive
3. Harder to interpret

### When to use:
Polynomial regression is preferred when the relationship between variables is nonlinear.
