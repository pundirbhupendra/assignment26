# Ridge Regression Assignment

## Q1. What is Ridge Regression, and how does it differ from ordinary least squares regression?

Ridge Regression is a regularized form of linear regression that adds an L2 penalty term to the loss function.

### Formula:
Loss = RSS + λΣβ²

### Difference from OLS:
- Ordinary Least Squares (OLS) minimizes only residual errors.
- Ridge Regression minimizes residual errors plus a penalty on large coefficients.
- Ridge helps reduce overfitting and handles multicollinearity better.

## Q2. What are the assumptions of Ridge Regression?

The assumptions of Ridge Regression are:

1. Linear relationship between variables
2. Independence of observations
3. Constant variance of errors
4. Normally distributed residuals
5. No perfect multicollinearity

Ridge Regression is especially useful when multicollinearity exists.

## Q3. How do you select the value of lambda in Ridge Regression?

The tuning parameter lambda (λ) is selected using:

1. Cross-validation
2. Grid Search
3. Validation datasets

### Interpretation:
- Small λ → behaves like linear regression
- Large λ → stronger regularization

The best lambda minimizes prediction error.

## Q4. Can Ridge Regression be used for feature selection?

Ridge Regression is not ideal for feature selection because it reduces coefficients but usually does not make them exactly zero.

Lasso Regression is preferred for feature selection since it can eliminate features completely.

## Q5. How does Ridge Regression perform in the presence of multicollinearity?

Ridge Regression performs well with multicollinearity.

### Reason:
When predictors are highly correlated:
- OLS coefficients become unstable
- Ridge shrinks coefficients and reduces variance

This improves model stability and prediction accuracy.

## Q6. Can Ridge Regression handle both categorical and continuous variables?

Yes, Ridge Regression can handle both categorical and continuous variables.

Categorical variables should first be converted into numerical form using:
- One-Hot Encoding
- Label Encoding

## Q7. How do you interpret the coefficients of Ridge Regression?

The coefficients represent the relationship between independent variables and the target variable.

### Important:
- Coefficients are shrunk toward zero due to regularization.
- Positive coefficients increase the target value.
- Negative coefficients decrease the target value.

## Q8. Can Ridge Regression be used for time-series data analysis?

Yes, Ridge Regression can be used for time-series forecasting.

### Usage:
Past observations and lag variables can be used as predictors.

### Benefits:
- Reduces overfitting
- Handles correlated predictors
- Improves forecasting stability
