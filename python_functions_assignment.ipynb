## Q1. Keyword used to create a function and return odd numbers from 1 to 25.

```
# 'def' keyword is used to create a function

def odd_numbers():
    return [i for i in range(1, 26) if i % 2 != 0]

print(odd_numbers())
```

## Q2. Demonstrate *args and **kwargs.

```
# Example of *args

def add_numbers(*args):
    return sum(args)

print("Using *args:", add_numbers(1, 2, 3, 4, 5))

# Example of **kwargs

def student_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")

print("\nUsing **kwargs:")
student_details(name="Shiva", age=25, course="Python")
```

## Q4. Generator function and yield keyword.

```
# Generator function uses yield keyword to return values one at a time.

def square_generator(n):
    for i in range(n):
        yield i ** 2

gen = square_generator(5)

for value in gen:
    print(value)
```

## Q5. Generator function for prime numbers less than 1000.

```
def prime_generator():
    for num in range(2, 1000):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num

prime_gen = prime_generator()

for _ in range(20):
    print(next(prime_gen))
```

## Q6. First 10 Fibonacci numbers using while loop.

```
a, b = 0, 1
count = 0

while count < 10:
    print(a)
    a, b = b, a + b
    count += 1
```

## Q8. Check palindrome using while loop.

```
number = int(input("Enter a number: "))

original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

## Q9. Print odd numbers from 1 to 100 using list comprehension.

```
numbers = [i for i in range(1, 101)]

odd_numbers = [num for num in numbers if num % 2 != 0]

print(odd_numbers)
```
