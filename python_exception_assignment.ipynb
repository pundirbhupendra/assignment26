## Q1. What is an Exception in Python? Difference between Exceptions and Syntax Errors.

```
# Exception:
# An exception is an error that occurs during the execution of a program.

# Difference:
# Syntax Error -> Occurs due to invalid syntax before execution.
# Exception -> Occurs during execution even if syntax is correct.

# Example of Syntax Error:
# print("Hello"

# Example of Exception:
a = 10
b = 0

try:
    print(a / b)
except ZeroDivisionError as e:
    print("Exception:", e)
```

## Q2. What happens when an exception is not handled?

```
# If an exception is not handled, the program terminates abnormally.

a = 10
b = 0

# This will stop the program
print(a / b)
```

## Q3. Statements used to catch and handle exceptions.

```
# try and except are used to catch and handle exceptions.

try:
    num = int("abc")
except ValueError as e:
    print("Handled Exception:", e)
```

## Q4. Explain try, else, finally, and raise.

```
try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Division Result:", result)

finally:
    print("This block always executes")

# raise example
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

## Q5. What are Custom Exceptions?

```
# Custom exceptions are user-defined exceptions.

# They are used to create meaningful error messages
# according to application requirements.

class InvalidAgeError(Exception):
    pass

age = -1

if age < 0:
    raise InvalidAgeError("Invalid age entered")
```

## Q6. Create and use a custom exception class.

```
class MarksError(Exception):
    def __init__(self, message):
        self.message = message
        super().__init__(self.message)

def check_marks(marks):
    if marks < 0 or marks > 100:
        raise MarksError("Marks should be between 0 and 100")
    else:
        print("Valid Marks")

try:
    check_marks(120)

except MarksError as e:
    print("Custom Exception Caught:", e)
```
