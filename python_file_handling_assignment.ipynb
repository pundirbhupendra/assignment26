## Q2. Why close() function is used?

```
# close() function is used to close the file.

file = open("sample.txt", "w")
file.write("Hello World")
file.close()

print("File closed successfully")

print("\nImportance of close():")
print("1. Saves changes properly")
print("2. Frees system resources")
print("3. Prevents data corruption")
```

## Q3. Create, write, close, open and read a text file.

```
# Creating and writing into file

file = open("data_scientist.txt", "w")

file.write("I want to become a Data Scientist")

file.close()

# Reading the file

file = open("data_scientist.txt", "r")

content = file.read()

print(content)

file.close()
```

## Q4. Explain read(), readline() and readlines().

```
# Create sample file

file = open("sample_read.txt", "w")

file.write("Line 1\nLine 2\nLine 3")

file.close()

# read()
file = open("sample_read.txt", "r")
print("Using read():")
print(file.read())
file.close()

# readline()
file = open("sample_read.txt", "r")
print("\nUsing readline():")
print(file.readline())
file.close()

# readlines()
file = open("sample_read.txt", "r")
print("\nUsing readlines():")
print(file.readlines())
file.close()
```

## Q5. Why with statement is used with open()?

```
# with statement automatically closes the file.

with open("with_example.txt", "w") as file:
    file.write("Using with statement")

with open("with_example.txt", "r") as file:
    print(file.read())

print("\nAdvantages:")
print("1. Automatically closes the file")
print("2. Better resource management")
print("3. Reduces chances of errors")
```

## Q6. Explain write() and writelines().

```
# write() example

with open("write_example.txt", "w") as file:
    file.write("Hello Python")

# writelines() example

lines = ["Line 1\n", "Line 2\n", "Line 3\n"]

with open("writelines_example.txt", "w") as file:
    file.writelines(lines)

# Reading files

with open("write_example.txt", "r") as file:
    print("Using write():")
    print(file.read())

with open("writelines_example.txt", "r") as file:
    print("\nUsing writelines():")
    print(file.read())
```
