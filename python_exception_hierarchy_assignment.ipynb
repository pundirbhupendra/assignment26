## Q1. Why use the Exception class while creating Custom Exceptions?

```
# Exception class is the base class for all exceptions in Python.

# Using Exception class helps:
# 1. Maintain proper exception hierarchy
# 2. Enable custom exceptions to behave like built-in exceptions
# 3. Allow exception handling using try-except blocks

class CustomError(Exception):
    pass

try:
    raise CustomError("This is a custom exception")
except CustomError as e:
    print(e)
```

## Q3. Errors defined in ArithmeticError class.

```
# ArithmeticError includes:
# 1. ZeroDivisionError
# 2. OverflowError
# 3. FloatingPointError

# Example of ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)

# Example of OverflowError
try:
    import math
    print(math.exp(1000))
except OverflowError as e:
    print("OverflowError:", e)
```

## Q4. Why LookupError class is used?

```
# LookupError is the base class for errors raised
# when a key or index is not found.

# Example of KeyError
try:
    data = {"name": "Shiva"}
    print(data["age"])
except KeyError as e:
    print("KeyError:", e)

# Example of IndexError
try:
    numbers = [1, 2, 3]
    print(numbers[5])
except IndexError as e:
    print("IndexError:", e)
```
