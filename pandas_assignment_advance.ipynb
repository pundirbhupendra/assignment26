## Q1. List any five functions of the pandas library with execution.

```
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Claire"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print("head():")
print(df.head())

print("\ntail():")
print(df.tail())

print("\nshape:")
print(df.shape)

print("\ncolumns:")
print(df.columns)

print("\ndescribe():")
print(df.describe())
```

## Q2. Re-index DataFrame starting from 1 and increment by 2.

```
import pandas as pd

df = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6],
    "C": [7, 8, 9]
})

df.index = range(1, len(df)*2, 2)

print(df)
```

## Q3. Calculate sum of first three values.

```
import pandas as pd

df = pd.DataFrame({
    "Values": [10, 20, 30, 40, 50]
})

sum_values = df["Values"].head(3).sum()

print("Sum of first three values:", sum_values)
```

## Q4. Create Word_Count column.

```
import pandas as pd

df = pd.DataFrame({
    "Text": ["Hello world", "Pandas is powerful", "Python programming"]
})

df["Word_Count"] = df["Text"].apply(lambda x: len(x.split()))

print(df)
```

## Q7. Extract username from email.

```
import pandas as pd

df = pd.DataFrame({
    "Email": ["john.doe@example.com", "alice@gmail.com"]
})

df["Username"] = df["Email"].apply(lambda x: x.split("@")[0])

print(df)
```

## Q8. Select rows based on conditions.

```
import pandas as pd

df = pd.DataFrame({
    "A": [3, 8, 6, 2, 9],
    "B": [5, 2, 9, 3, 1],
    "C": [1, 7, 4, 5, 2]
})

result = df[(df["A"] > 5) & (df["B"] < 10)]

print(result)
```

## Q9. Calculate mean, median, and standard deviation.

```
import pandas as pd

df = pd.DataFrame({
    "Values": [10, 20, 30, 40, 50]
})

print("Mean:", df["Values"].mean())
print("Median:", df["Values"].median())
print("Standard Deviation:", df["Values"].std())
```

## Q10. Create MovingAverage column.

```
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range(start="2023-01-01", periods=10),
    "Sales": [100, 120, 130, 140, 150, 160, 170, 180, 190, 200]
})

df["MovingAverage"] = df["Sales"].rolling(window=7).mean()

print(df)
```

## Q11. Create Weekday column.

```
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range(start="2023-01-01", periods=5)
})

df["Weekday"] = df["Date"].dt.day_name()

print(df)
```

## Q12. Select rows between two dates.

```
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range(start="2023-01-01", periods=40)
})

filtered_df = df[(df["Date"] >= "2023-01-01") & (df["Date"] <= "2023-01-31")]

print(filtered_df)
```
