## Q3. Explain run(), start(), join(), isAlive().

```
import threading
import time

def task():
    print("Thread is running")
    time.sleep(2)
    print("Thread execution completed")

thread = threading.Thread(target=task)

# start()
thread.start()

# join()
thread.join()

# is_alive()
print("Is thread alive?", thread.is_alive())

# run()
# run() method contains the code executed by the thread.
```

## Q4. Create two threads to print squares and cubes.

```
import threading

numbers = [1, 2, 3, 4, 5]

def print_squares():
    print("Squares:")
    for i in numbers:
        print(i ** 2)

def print_cubes():
    print("Cubes:")
    for i in numbers:
        print(i ** 3)

thread1 = threading.Thread(target=print_squares)
thread2 = threading.Thread(target=print_cubes)

thread1.start()
thread2.start()

thread1.join()
thread2.join()
```

## Q6. Explain deadlocks and race conditions.

```
# Deadlock:
# A deadlock occurs when two or more threads wait indefinitely
# for resources locked by each other.

# Race Condition:
# A race condition occurs when multiple threads access and modify
# shared data simultaneously, causing unexpected results.

import threading

counter = 0

def increment():
    global counter
    for _ in range(100000):
        counter += 1

thread1 = threading.Thread(target=increment)
thread2 = threading.Thread(target=increment)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("Counter Value:", counter)
```
